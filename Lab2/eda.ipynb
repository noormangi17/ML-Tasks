```python
import numpy as np 
import pandas as pd
df=pd.read_csv("dataset.csv")
customers=df.shape[0]
print("Number of customers: ",customers)
print("Number of features: ",len(df.columns))
print("Number of numerical features: ",df.select_dtypes(include='number').columns)
print("Number of categorical features: ",df.select_dtypes(include='object').columns)
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")
print("Number of numerical features: ",df.select_dtypes(include='number').columns)
print("Number of categorical features: ",df.select_dtypes(include='object').columns)
```

```python
df["Churn"]=df["Churn"].map({
    "Yes":1,
    "No":0
})
print(df["Churn"])
```

```python
y=df["Churn"]
x=df.drop("Churn",axis=1)
```

```python
x=pd.get_dummies(x,drop_first=True)
```

```python

x["TotalCharges"]=x["TotalCharges"].fillna(x["TotalCharges"].mean())
print(x.dtypes)
print(x.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42,stratify=y
)
```

```python
print("Train churn %:", y_train.mean() * 100)
print("Test churn %:", y_test.mean() * 100)
```
